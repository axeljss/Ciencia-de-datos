
# Diccionarios

Guardan informacion en pares clave-valor. son ideales para repressentar un registro.

```python
pasajero = {
    "nombre" : "Axel Jasso",
    "edad" : 20,
    "clase" : 3,
    "Tarifa": 7.25,
    "sobrevivio": False
}

print(pasajero["nombre"])
print(pasajero["edad"])
print(pasajero)

#agregar nueva clave
pasajero["puerto_embarque"] = "Southampton"
print(pasajero)
```


# Funciones



```python
def clasificar_edad(edad):
    """Recibe una edad y regresa la categoria correspondiente."""
    if edad < 12:
        return "niño"
    elif edad < 18:
        return "adolescente"
    elif edad < 65:
        return "adulto"
    else:
        return "adulto mayor"

#probamos la funcion con distintos valores
for edad in [5, 15, 40, 80]:
    print(edad, "->", clasificar_edad(edad))
```


Crear funcion que permita agregar a una lista un diccionario hasta un maximo de 4 pasajeros. usando if, def, for. Que se pregunte nombre, edad, clase, tarifa. Y que diga cuantos espacios quedan mientras vas agregando.

```python
def agregar_pasajeros():
    """Permite registrar hasta 4 pasajeros en una lista."""
    lista_nueva = []
    maximo_pasajeros = 4
    
    for i in range(maximo_pasajeros):
        espacios_quedan = maximo_pasajeros - len(lista_nueva)
        if espacios_quedan > 0:
            print("quedan ", espacios_quedan,  "espacios disponibles")
            
            #datos
            nombre = input("Ingresa el nombre del pasajero: ")
            edad = int(input("Ingresa la edad: "))
            clase = int(input("Ingresa la clase (1, 2 o 3): "))
            tarifa = float(input("Ingresa la tarifa: "))
            
            #Dic
            nuevo_pasajero = {
                "nombre": nombre,
                "edad": edad,
                "clase": clase,
                "Tarifa": tarifa
            }
            #agregar 
            lista_nueva.append(nuevo_pasajero)
            
            print("Se agrego pasajero: ", nuevo_pasajero["nombre"])
            
            #preguntar si agrega mas
            if maximo_pasajeros - len(lista_nueva) > 0:
                continuar = input("¿Deseas agregar otro pasajero? s para si, n para no ")
                if continuar.lower() != 's':
                    break
                    
    #Mostrar
    if len(lista_nueva) == maximo_pasajeros:
        print("se lleno la lista")
        
    return lista_nueva

#Prueba
pasajeros_registrados = agregar_pasajeros()

print("lista de pasajeros: ")
for p in pasajeros_registrados:
    print(p)
```
